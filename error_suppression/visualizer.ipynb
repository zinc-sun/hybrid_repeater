```python
import numpy as np
import matplotlib.pyplot as plt
plt.rcParams["font.size"] = 14.0
%matplotlib inline

plt.rcParams.update({
    "text.usetex": True,
    "font.family": "serif",
    "font.serif": ["Computer Modern Roman"],
    "font.size": 14.0
})
```

```python
local_loss_lst = np.linspace(0, 0.9, 100)

fid_lst_raw = np.load("fid_lst_raw.npy", allow_pickle = True)
fid_lst_pnr = np.load("fid_lst_pnr.npy", allow_pickle = True)
fid_lst_ED = np.load("fid_lst_ED.npy", allow_pickle = True)
fid_lst_pnr_ED = np.load("fid_lst_pnr_ED.npy", allow_pickle = True)
fid_lst_reemission = np.load("fid_lst_reemission.npy", allow_pickle = True)
# fid_lst_pnr_reemission_flip = np.load("fid_lst_pnr_reemission_flip.npy", allow_pickle = True)
fid_lst_pnr_reemission = np.load("fid_lst_pnr_reemission.npy", allow_pickle = True)
```

```python
plt.title("Fidelity with different error suppression")
plt.xlabel("Local loss")
plt.ylabel("F")
plt.plot(local_loss_lst, fid_lst_raw, label = 'raw')
plt.plot(local_loss_lst, fid_lst_pnr, label ='PNR')
plt.plot(local_loss_lst, fid_lst_ED, label = 'EPL')
plt.plot(local_loss_lst, fid_lst_pnr_ED, label = 'PNR+EPL')
plt.plot(local_loss_lst, fid_lst_reemission, label = 'RE')
# plt.plot(local_loss_lst, fid_lst_pnr_reemission_flip, label ='PNR+re-emission (flip)')
plt.plot(local_loss_lst, fid_lst_pnr_reemission, label ='PNR+RE')
plt.legend(loc='best')
# plt.yscale('log')
plt.savefig('fidelity_error_suppression.pdf', bbox_inches='tight')
```
