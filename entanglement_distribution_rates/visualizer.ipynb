```python
import numpy as np
import matplotlib.pyplot as plt
plt.rcParams.update({
    "text.usetex": False,
    "font.family": "serif",
    "mathtext.fontset": "cm",
    "font.size": 14
})
%matplotlib inline
```

```python
rates_2hop_atom = np.load('rates_2hop_atom_90.npy', allow_pickle = True)
rates_3hop_atom = np.load('rates_3hop_atom_90.npy', allow_pickle = True)
rates_4hop_atom = np.load('rates_4hop_atom_90.npy', allow_pickle = True)
rates_2hop_hybrid = np.load('rates_2hop_hybrid_f95_l_50_n_40.npy', allow_pickle = True)
rates_3hop_hybrid = np.load('rates_3hop_hybrid_f95_l_50_n_40.npy', allow_pickle = True)
rates_4hop_hybrid = np.load('rates_4hop_hybrid_f95_l_50_n_40.npy', allow_pickle = True)
```

```python
distances_20 = np.linspace(0.1, 1000, 20)
distances_40 = np.linspace(0.1, 1000, 40)
distances_500 = np.linspace(0.1, 1000, 500)
```

```python
plt.title(r"Entanglement distribution rates")
plt.xlabel('Distance (km)')
plt.ylabel('Entanglement distribution rate (Hz)')

plt.plot(distances_20, rates_2hop_atom, label = 'EL (atom)', linestyle = 'dashed', color = 'red')
plt.plot(distances_20, rates_3hop_atom, label = '1-hop (atom)', linestyle = 'dashed', color = 'orange')
plt.plot(distances_20, rates_4hop_atom, label = '2-hop (atom)', linestyle = 'dashed', color = 'blue')

plt.plot(distances_40, rates_2hop_hybrid, label = 'EL (hybrid)', color = 'red')
plt.plot(distances_40, rates_3hop_hybrid, label = '1-hop (hybrid)', color = 'orange')
plt.plot(distances_40, rates_4hop_hybrid, label = '2-hop (hybrid)', color = 'blue')

plt.ylim(bottom=10**-8)
# plt.xlim(0, 300)
plt.yscale('log')
plt.legend(loc = 'best')
plt.savefig('entanglement_rates_50.pdf', bbox_inches='tight')
```
