```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

plt.rcParams.update({
    "text.usetex": False,
    "font.family": "serif",
    "mathtext.fontset": "cm",
    "font.size": 14
})
```

```python
sk_rates_2hop_atom = np.load('sk_rates_2hop_atom_50.npy', allow_pickle = True)
sk_rates_3hop_atom = np.load('sk_rates_3hop_atom_50.npy', allow_pickle = True)
sk_rates_4hop_atom = np.load('sk_rates_4hop_atom_50.npy', allow_pickle = True)
sk_rates_2hop_hybrid = np.load('sk_rates_2hop_hybrid_50.npy', allow_pickle = True)
sk_rates_3hop_hybrid = np.load('sk_rates_3hop_hybrid_50.npy', allow_pickle = True)
sk_rates_4hop_hybrid = np.load('sk_rates_4hop_hybrid_50.npy', allow_pickle = True)
```

```python
# distances = np.linspace(0.1, 400, 20)
distances = np.linspace(0.1, 1000, 20)
```

```python
plt.title(r"BB84 Secret key rates")
plt.xlabel('Distance (km)')
plt.ylabel('Secret key rates (Hz)')

plt.plot(distances, sk_rates_2hop_atom, label = 'EL (atom)', linestyle = 'dashed', color = 'red')
plt.plot(distances, sk_rates_3hop_atom, label = '1-hop (atom)', linestyle = 'dashed', color = 'orange')
plt.plot(distances, sk_rates_4hop_atom, label = '2-hop (atom)', linestyle = 'dashed', color = 'blue')

plt.plot(distances, sk_rates_2hop_hybrid, label = 'EL (hybrid)', color = 'red')
plt.plot(distances, sk_rates_3hop_hybrid, label = '1-hop (hybrid)', color = 'orange')
plt.plot(distances, sk_rates_4hop_hybrid, label = '2-hop (hybrid)', color = 'blue')

plt.ylim(bottom=10**-8)
# plt.xlim(0, 300)
plt.yscale('log')
plt.legend(loc = 'best')
plt.savefig('key_rates_50.pdf', bbox_inches='tight')
```
